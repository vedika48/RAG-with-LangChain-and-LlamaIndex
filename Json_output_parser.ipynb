```python
#can stream partial JSON responses in real-time, allowing users to see the output being built step by step from an empty result to complete answer
from langchain_openai import OpenAI
from langchain.output_parsers.json import SimpleJsonOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
from dotenv import load_dotenv
load_dotenv()
model = OpenAI(model_name="gpt-3.5-turbo-instruct", temperature=0.0)
```

```python
from langchain.output_parsers.json import SimpleJsonOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
json_prompt = PromptTemplate.from_template(
    "Return JSON object with an answer key that answers the following questions: {question}"
)
json_parser = SimpleJsonOutputParser()
json_chain = json_prompt | model | json_parser
```

```python
#Streamed output with parallel JSON
print(list(json_chain.stream({"question":"Who discovered"})))
#Expected output
# [{},{'answer':''},{'answer':'A'},{'answer':'Al'},{'answer':'Ale'},{'answer':'Alex'},{'answer':'Alexa'},{'answer':'Alexan'},{'answer':'Alexander'}]
```
