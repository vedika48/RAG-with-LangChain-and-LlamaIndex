```python
from langchain_core.prompts import FewShotPromptTemplate
from langchain_core.prompts import PromptTemplate
```

```python
#Define a template
exp_template = PromptTemplate(
    input_variables=["question", "answer"],
    template="Q: {question}\nA: {answer}"
)
```

```python
#Define some example
examples = [
    {"question": "Who was the first", "answer": "George"},
    {"question": "What year", "answer": "1939"}
]
```

```python
# Define the prompt with few examples and placeholhers for dynamic input
few_shot_template = FewShotPromptTemplate(
    examples = examples,
    example_prompt = exp_template,
    prefix = "Here are some facts:",
    suffix = "Q: {user_question}\nA:",
    input_variables = ["user_question"] #Dynamic part
)
```

```python
temp = few_shot_template.invoke({"user_question":"Who wrote it"})
print(temp)
```
