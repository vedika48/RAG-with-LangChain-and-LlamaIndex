```python
# Chat model where string input is automatically converted to HummanMessage
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage
```

```python
chat = ChatOllama(
    model = "llama3.2",
    temperature = 0.7,
    max_token = 100,
    stop = ["\n"],
)
```

```python
response = chat.invoke("What is the capital of USA")
print(response.content)
```
