```python
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, SystemMessage
```

```python
@tool
def weather_func(location: str) -> str:
    """Get the current weather information"""
    return f"The weather in {location} is 25F."
```

```python
chat = ChatOllama(model="llama3.2",)
llm_with_tools = chat.bind_tools([weather_func])
conversation = [
    SystemMessage(content="You are a chatbot thet can interact with tools. Use my tool weather_func."),
    HumanMessage(content="What's the weather like in Pune?"),
]
ai_msg = llm_with_tools.invoke(conversation)
messages = []
```

```python
for tool_call in ai_msg.tool_calls:
    selected_tools = {"weather_func": weather_func}[tool_call["name"].lower()]
    tool_msg = selected_tools.invoke(tool_call)
    messages.append(tool_msg)

print(messages)
```
