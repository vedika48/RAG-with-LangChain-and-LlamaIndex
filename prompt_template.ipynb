```python
from langchain_core.prompts import PromptTemplate
```

```python
# Define a prompt template with placeholder for dynamic input
prompt_template = PromptTemplate(
    input = ["topic"],
    template = "Explain the key feature of {topic}."
)
```

```python
#invoke the template
prompt_template.invoke({"topic" : "RAG"})
```

```python
prompt_template = PromptTemplate(
    input = ["username", "topic", "level"],
    template = "Hello {username}, please provide a {level}-level summary of {topic}."
)
```

```python
template = prompt_template.format(username="Vedika", topic="RAG", level="beginner")
print(template)
```
