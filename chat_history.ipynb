```python
from langchain_community.chat_message_histories import SQLChatMessageHistory
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
def get_session_history(session_id):
    return SQLChatMessageHistory(session_id=session_id, connection="sqlite:///memory.db",)

model = ChatOllama(model="llama3.2")
runnable_with_history = RunnableWithMessageHistory(model, get_session_history,)
```

```python
runnable_with_history.invoke(
    [HumanMessage(content="hi, I'm Vedika")],
    config={"configurable":{"session_id":"1"}},
)
runnable_with_history.invoke(
    [HumanMessage(content="Can you recommend a good restaurant in Pune?")],
    config={"configurable":{"session_id":"1"}},
)
```

```python
response = runnable_with_history.invoke(
    [HumanMessage(content="What's the weather like here and do you remember my name?")],
    config={"configurable":{"session_id":"1"}},
)
print(response.content)
```
